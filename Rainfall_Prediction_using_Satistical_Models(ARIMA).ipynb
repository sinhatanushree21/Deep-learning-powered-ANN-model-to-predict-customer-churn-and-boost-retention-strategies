**Time series analysis**

check for data is stationary or non stationary

````python
plt.figure(figsize=(14,5))
plt.plot(df.index, df['precipitation(mm/day)'])
plt.title("Daily Precipitation Time Series")
plt.xlabel("Year")
plt.ylabel("Rainfall (mm)")
plt.show()
````

````python
rolling_mean = df['precipitation(mm/day)'].rolling(window=365).mean()
rolling_std = df['precipitation(mm/day)'].rolling(window=365).std()

plt.figure(figsize=(14,6))
plt.plot(df['precipitation(mm/day)'], label='Original', alpha=0.5)
plt.plot(rolling_mean, label='Rolling Mean (1 Year)', color='red')
plt.plot(rolling_std, label='Rolling Std (1 Year)', color='black')
plt.legend()
plt.title("Rolling Mean & Standard Deviation")
plt.show()
````

The precipitation time series is non-stationary

confirm statistically using ADF:

null hypothesis : data is non-stationary (unit root)

if p value < 0.05 reject null hypothesis (means data is stionary)
if p value > 0.05 accept null hypothesis (fail to reject null hypothesis) (means data is non-stationary)




````python
from statsmodels.tsa.stattools import adfuller

adf_test = adfuller(df['precipitation(mm/day)'].dropna())

print('ADF Test results')
print(f'ADF Statistic: {round(adf_test[0],3)}')
print(f'p-value: {round(adf_test[1],3)}')
print('critical_values')
for key, value in adf_test[4].items():
    print(f'{key}: {round(value,3)}')
````

````python
#ADF on Rolling mean
from statsmodels.tsa.stattools import adfuller

adf_test = adfuller(rolling_mean.dropna())

print('ADF Test results')
print(f'ADF Statistic: {round(adf_test[0],3)}')
print(f'p-value: {round(adf_test[1],3)}')
print('critical_values')
for key, value in adf_test[4].items():
    print(f'{key}: {round(value,3)}')
````

**AR test**

````python
train_size = int(len(monthly_max)* 0.70)

train_data = monthly_max[0:train_size]
test_data = monthly_max[train_size:]
````

````python
from statsmodels.tsa.ar_model import AutoReg
from sklearn.metrics import mean_squared_error
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
````

**AR**

````python
# Fit AR model on training data

model = AutoReg(train_data, lags=45)
model_fit = model.fit()
````

````python
model = AutoReg(train_data, lags=45)
model_fit = model.fit()


# Make predictions on test data


predictions = model_fit.predict(
    start=len(train_data),
    end=len(train_data) + len(test_data) - 1,
    dynamic=False
)

# Ensure predictions are non-negative for RMSLE and SI calculation
predictions_clipped = np.maximum(0, predictions)


 # Plot actual vs predicted

plt.figure(figsize=[10, 5])
plt.plot(test_data.index, test_data, label='Test data')
plt.plot(test_data.index, predictions, color='red', linestyle='--', label='Predicted Rainfall')
plt.title('AR model')
plt.xlabel('Year')
plt.ylabel('Rainfall (in mm)')
plt.legend()
plt.show()


# Evaluate model performance
from sklearn.metrics import mean_absolute_error, mean_squared_log_error, r2_score

rmse = round(np.sqrt(mean_squared_error(test_data, predictions)), 2)
mae = round(mean_absolute_error(test_data, predictions), 2)
rmsle = round(np.sqrt(mean_squared_log_error(test_data, predictions_clipped)), 2)
r2 = round(r2_score(test_data, predictions), 2)
si = round(np.sqrt(mean_squared_error(test_data, predictions)) / np.mean(test_data), 2)
nse = round(1 - (np.sum((test_data - predictions)**2) / np.sum((test_data - np.mean(test_data))**2)), 2)

print('RMSE:', rmse)
print('MAE:', mae)
print('RMSLE:', rmsle)
print('R2 Score:', r2)
print('Scatter Index (SI):', si)
print('Nash-Sutcliffe Efficiency (NSE):', nse)
````

**MA Model**

````python
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error, mean_squared_log_error, r2_score


# Fit the Moving Average (MA) model
# order=(p,d,q) → here p=0 (no AR), d=0 (no differencing), q=30 (MA terms)
model = ARIMA(train_data, order=(0, 0,90))
model_fit = model.fit()


# Make predictions on the test data
predictions = model_fit.predict(
    start=len(train_data),
    end=len(train_data) + len(test_data) - 1,
    dynamic=False
)

# Ensure predictions are non-negative for RMSLE and SI calculation
predictions_clipped = np.maximum(0, predictions)


# Plot actual vs predicted values
plt.figure(figsize=(10, 5))
plt.plot(test_data.index, test_data, label='Test data')
plt.plot(test_data.index, predictions, color='red', linestyle='--', label='Predicted Rainfall')
plt.title('MA Model')
plt.xlabel('Year')
plt.ylabel('Max Rainfall (in mm)')
plt.legend()
plt.show()

#model evaluation
rmse = round(np.sqrt(mean_squared_error(test_data, predictions)),2)
mae = round(mean_absolute_error(test_data, predictions), 2)
#mape = round(mean_absolute_percentage_error(test_data, predictions), 2) # Removed MAPE
rmsle = round(np.sqrt(mean_squared_log_error(test_data, predictions_clipped)), 2)
r2 = round(r2_score(test_data, predictions), 2)
si = round(np.sqrt(mean_squared_error(test_data, predictions)) / np.mean(test_data), 2)
nse = round(1 - (np.sum((test_data - predictions)**2) / np.sum((test_data - np.mean(test_data))**2)), 2)

print(f'Root Mean Square Error : {rmse}')
print(f'Mean Absolute Error: {mae}')
#print(f'Mean Absolute Percentage Error: {mape}') # Removed MAPE
print(f'Root Mean Squared Logarithmic Error: {rmsle}')
print(f'R2 Score: {r2}')
print(f'Scatter Index (SI): {si}')
print(f'Nash-Sutcliffe Efficiency (NSE): {nse}')
````
